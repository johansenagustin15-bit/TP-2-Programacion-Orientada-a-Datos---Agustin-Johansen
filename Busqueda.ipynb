# Búsqueda de películas

Este cuaderno importa las funciones desde `peliculas.py`, carga y valida los datos, y ejecuta el menú interactivo. Ejecutá las celdas en orden. Elegí un formato (`csv` o `json`) en la primera celda.

```python
from pathlib import Path

from peliculas import (
    cargar_peliculas,
    inspeccionar_y_validar,
    limpiar_y_transformar,
    menu_busqueda,
)

FORMATO_ENTRADA = "csv"  # Cambiar a "json" para leer el otro formato.
rutas_datos = {
    "csv": Path("data") / "IMDb 100 Mejores Peliculas.csv",
    "json": Path("data") / "IMDb 100 Mejores Peliculas.json",
}

if FORMATO_ENTRADA not in rutas_datos:
    raise ValueError('FORMATO_ENTRADA debe ser "csv" o "json".')

datos_originales = cargar_peliculas(rutas_datos[FORMATO_ENTRADA])
datos_validados = inspeccionar_y_validar(datos_originales, FORMATO_ENTRADA.upper())
peliculas = limpiar_y_transformar(datos_validados)
print("Columnas disponibles:", peliculas.columns.tolist())
display(peliculas.head())
```

```python
menu_busqueda(peliculas)
```
