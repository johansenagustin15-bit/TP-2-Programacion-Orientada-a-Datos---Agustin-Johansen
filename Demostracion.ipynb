# IMDb: 100 mejores películas

Este cuaderno documenta el flujo reproducible del proyecto: **leer → inspeccionar y validar → limpiar y transformar → analizar → comunicar resultados**. Las funciones están definidas en `peliculas.py` para no duplicar código entre este trabajo y la aplicación de búsqueda.

El CSV y el JSON tienen las mismas películas. El JSON se generó desde el CSV para demostrar que se pueden leer ambos formatos.

Ejecutá las celdas en orden. En `Busqueda.ipynb` está el menú interactivo para que una persona elija sus propios filtros.

```python
from pathlib import Path
from pandas.testing import assert_frame_equal

from peliculas import (
    buscar_peliculas,
    cargar_peliculas,
    inspeccionar_y_validar,
    limpiar_y_transformar,
)

RUTA_CSV = Path("data") / "IMDb 100 Mejores Peliculas.csv"
RUTA_JSON = Path("data") / "IMDb 100 Mejores Peliculas.json"

peliculas_csv_originales = cargar_peliculas(RUTA_CSV)
peliculas_json_originales = cargar_peliculas(RUTA_JSON)

print("CSV:", peliculas_csv_originales.shape)
print("JSON:", peliculas_json_originales.shape)
print("Columnas de entrada:", peliculas_csv_originales.columns.tolist())
display(peliculas_csv_originales.head())
```

```python
csv_validado = inspeccionar_y_validar(peliculas_csv_originales, "CSV")
json_validado = inspeccionar_y_validar(peliculas_json_originales, "JSON")

peliculas_csv = limpiar_y_transformar(csv_validado)
peliculas_json = limpiar_y_transformar(json_validado)
assert_frame_equal(peliculas_csv, peliculas_json, check_dtype=False)

print("Los datos de CSV y JSON son equivalentes después de la limpieza.")
print("Columnas finales:", peliculas_csv.columns.tolist())
display(peliculas_csv.head())
```

```python
# Ejemplo reproducible de análisis. La aplicación permite ingresar otros filtros.
CRITERIO = "genero"
VALOR = "Drama"

resultado_csv = buscar_peliculas(peliculas_csv, CRITERIO, VALOR)
resultado_json = buscar_peliculas(peliculas_json, CRITERIO, VALOR)
assert_frame_equal(resultado_csv, resultado_json, check_dtype=False)

print(f"Búsqueda de ejemplo: {CRITERIO} = {VALOR!r}")
print(f"Películas encontradas: {len(resultado_csv)}")
display(resultado_csv)
```
